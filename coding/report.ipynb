```python
# Consistency Analysis
df_Consistencyllama = pd.read_csv('ConsistencyLamma.csv')
df_ConsistencyGPT = pd.read_csv('ConsistencyGPT.csv')
df_ConsistencyHaiku = pd.read_csv('consistencyHaiku.csv')
```

```python
def hitung_konsistensi(df, nama_model):
    # 1. Kelompokkan (groupby) berdasarkan kolom delete_reason
    grouped = df.groupby('delete_reason')['Final label']
    
    # Menghitung total kemunculan kalimat tersebut
    total_muncul = grouped.count()
    
    # 2. CARA AMAN TANPA KURUNG SIKU: 
    # Gunakan .idxmax() untuk mengambil nama label terbanyak
    label_mayoritas = grouped.apply(lambda x: x.value_counts().idxmax())
    
    # Gunakan .max() untuk mengambil jumlah dari label terbanyak
    jumlah_mayoritas = grouped.apply(lambda x: x.value_counts().max())
    
    # 3. Hitung persentase konsistensi
    persentase = (jumlah_mayoritas / total_muncul) * 100
    
    # Menggabungkan hasil ke dalam DataFrame
    df_hasil = pd.DataFrame({
        'Model': nama_model,
        'Total Kemunculan': total_muncul,
        'Label Mayoritas': label_mayoritas,
        'Konsistensi (%)': persentase
    }).reset_index()
    
    return df_hasil
```

```python
# Menjalankan fungsi untuk masing-masing model
hasil_llama = hitung_konsistensi(df_Consistencyllama, 'Llama')
hasil_gpt = hitung_konsistensi(df_ConsistencyGPT, 'GPT')
hasil_haiku = hitung_konsistensi(df_ConsistencyHaiku, 'Haiku')
```

```python
# Menggabungkan seluruh hasil menjadi satu tabel perbandingan
df_perbandingan_konsistensi = pd.concat([hasil_gpt, hasil_llama, hasil_haiku], ignore_index=True)

# Menampilkan hasil akhirnya
print("=== HASIL UJI KONSISTENSI MODEL ===")
print(df_perbandingan_konsistensi.to_string())
```
